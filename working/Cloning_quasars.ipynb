```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import uniform
```

```python
z= np.loadtxt('dr7qso.dat',skiprows=80, usecols=3) #this is the redshift distribution of quasars
```

```python
counts, bins, patches = plt.hist(z,bins=100,density=True,histtype='step',lw=2);
plt.xlabel('z')
```

```python
plt.plot(bins[:-1], counts)
```

# Rejection sampling


As the proposal $q(x)$ I choose a uniform distribution between 0 and 5.5. Let's generate a random sample from this distribution:


```python
hx = bins[:-1]
N = 20000
max_hx = max(hx)
xi_sample = uniform.rvs(loc=min(z), scale=max(z), size=N)
u_sample = uniform.rvs(loc=0, scale=max_hx, size=N)
```

```python
vals_kept = []

for i in range(N):
    for j in range(len(hx)-1):
        if u_sample[i] <= hx[j+1] and u_sample[i] > hx[j]:
            vals_kept.append(xi_sample[i])
```

```python
plt.hist(vals_kept, bins=100, density=True)
```
