```python
import pandas as pd
from pathlib import Path

# ---- 1) Point to your file ----
dat_path = Path("dr7qso.dat")


columns = [
    # name                  start  end   note 
    ("designation",            1,   18),  # A18 DR7 Object Designation
    ("ra_deg",                19,   29),  # F11.6 RA (deg)
    ("dec_deg",               30,   40),  # F11.6 Dec (deg)
    ("z",                     41,   47)]


colspecs = [(start-1, end) for (_, start, end) in columns]
names    = [name for (name, _, _) in columns]


df = pd.read_fwf(
    dat_path,
    names=names,
    colspecs=colspecs
)

df.shape, df.head(3), df.info()
```
