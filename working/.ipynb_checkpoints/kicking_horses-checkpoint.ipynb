```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats
```

```python
def poisson(x, mu):
    return scipy.stats.poisson(mu).pmf(x)
```

```python
#10 groups observed for 20 years

deaths = np.array([0, 1, 2, 3, 4])
groups = np.array([109, 65, 22, 3, 1])

tot = np.sum(groups)
frequency = groups/tot
calc_mu = np.average(deaths, weights=frequency)

xarr = np.arange(0, 5)
yarr = poisson(xarr,calc_mu)

plt.scatter(deaths, frequency)
plt.plot(xarr, yarr, color='orange')
```
