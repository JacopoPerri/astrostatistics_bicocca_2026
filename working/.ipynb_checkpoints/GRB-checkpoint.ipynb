```python
import numpy as np
import sklearn
from sklearn.cluster import KMeans
from sklearn import preprocessing
import matplotlib.pyplot as plt
from astropy.io import ascii
from sklearn.model_selection import GridSearchCV
from sklearn.cluster import MeanShift
from sklearn.neighbors import KernelDensity
```

# Helpful functions

```python
def take_out_invalid(a, b):
    a_new = []
    b_new = []
    for i in range(len(a)):
        if a[i] != -999 and b[i] != -999 and a[i] >0 and b[i]>0:
            a_new.append(a[i])
            b_new.append(b[i])

    return a_new, b_new
```

```python
def plot_results(data, labels, log=False):
    lbl_set = set(labels)
    
    for lbl in lbl_set:
        x = data[:, 0]
        y = data[:, 1]

        x_lbl = x[labels == lbl]
        y_lbl = y[labels == lbl]

        plt.scatter(x_lbl, y_lbl, label='Group '+str(lbl), alpha=0.3)
        plt.legend()
        
        if log == True:
            plt.xscale('log')
            plt.yscale('log')
    
    plt.xlabel('log(t)')
    plt.ylabel('log(fl)')
```

# The dataset

```python
dataset = ascii.read('Summary_table.txt')
dataset
```

```python
t90 = dataset['col7']
flux = dataset['col10']

t90, flux = take_out_invalid(t90, flux)
```

```python
log_t = np.log(t90)
log_fl = np.log(flux)
```

```python
plt.scatter(log_t, log_fl, alpha=0.5, s=5)
plt.xlabel('T90 [s]')
plt.ylabel('Fluence [erg/cm^2]')
```

```python
plt.hist(log_t, bins=100)
```

```python
plt.hist(log_fl, bins=100)
```

```python
X = np.array([log_t, log_fl], np.newaxis).T
nsamp = len(log_t)
X.shape
```

# KMeans clustering with 2 clusters

```python
clf = KMeans(n_clusters=2,n_init='auto') 
clf.fit(X)

centers = clf.cluster_centers_ 
labels = clf.predict(X)
```

```python
plot_results(X, labels)
plt.scatter(centers[0, 0], centers[0, 1], color='black')
plt.scatter(centers[1, 0], centers[1, 1], color='black')
```

# Checking for multiple populations

what if there are actually more populations? let's use mean shift clustering
to both optimize the number of possible populations using a kde and find the populations
but firts we need to optimize the bandwidth

```python
X_logt = log_t[:, np.newaxis]
X_logfl = log_fl[:, np.newaxis]
```

```python
k_grid = np.linspace(0.1, 1, 20)

kde = KernelDensity()
gsearch_t = GridSearchCV(kde, {'bandwidth': k_grid})
gsearch_t.fit(X_logt)
bw_opt_t = gsearch_t.best_params_['bandwidth']
```

```python
gsearch_f = GridSearchCV(kde, {'bandwidth': k_grid})
gsearch_f.fit(X_logfl)
bw_opt_f = gsearch_f.best_params_['bandwidth']

print(bw_opt_t, bw_opt_f)
```

```python
scaler = preprocessing.StandardScaler()
bw = bw_opt_f
ms = MeanShift(bandwidth=bw, 
               bin_seeding=True)
ms.fit(scaler.fit_transform(X))
```

```python
labels = ms.labels_
lbl_set = set(labels)
n_clust = len(lbl_set) #whatever atp...
centers = ms.cluster_centers_

for lbl in lbl_set:
    x = X[:, 0]
    y = X[:, 1]

    x_lbl = x[labels == lbl]
    y_lbl = y[labels == lbl]

    plt.scatter(x_lbl, y_lbl, label='Group'+str(lbl), alpha=0.3)
    plt.legend()
```
