```python
import numpy as np
import pandas as pd
import urllib.request
from sklearn.preprocessing import LabelEncoder
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

# The dataset

```python
urllib.request.urlretrieve("https://raw.githubusercontent.com/nshaud/ml_for_astro/main/stars.csv", "stars.csv")
```

```python
df_stars = pd.read_csv('stars.csv')
df_stars
```

```python
le = LabelEncoder()
df_stars['Star type'] = le.fit_transform(df_stars['Star type'])
labels = le.inverse_transform(df_stars['Star type'])
class_names = le.classes_
print(class_names)
```

```python
T = df_stars['Temperature (K)']
L = df_stars['Luminosity(L/Lo)']
R = df_stars['Radius(R/Ro)']
absmag = df_stars['Absolute magnitude(Mv)']
```

```python
fig, ax = plt.subplots()
sns.scatterplot(data=df_stars, x='Temperature (K)', y='Luminosity(L/Lo)', hue=labels, palette='tab10')
plt.xscale('log')
plt.yscale('log')
ax.invert_xaxis()
```

# PCA

```python
X = np.array([T, L, R, absmag]) #4 features: temperature, luminosity, radius, absolute magnitude
X = X.T #shape = n_points, n_features
```

```python
# before doing the pca it's better to rescale the data
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
# can this be described with mean + 2 principal components?
pca = PCA(n_components=2) 
pca.fit(X) 
```

```python
eigenvalues = pca.transform(X) # compute the projection on the new smaller space: n_points, each of them is 2-dimensional
mean = pca.mean_ 
eigenvectors = pca.components_
```

```python
sns.scatterplot(x=eigenvalues[:,0], y=eigenvalues[:,1], hue=labels)
```
